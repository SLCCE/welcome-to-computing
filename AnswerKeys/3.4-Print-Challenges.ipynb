**Example 2:** Use print statements and escape sequences to produce the following output:

Engineering

>>>>is

>>>>>>>>Awesome


```python
print("Engineering")
print("\tis")
print("\t\tAwesome")
# Here we need both \n for newline and \t for tabs
print("Engineering\n\tis\n\t\tAwesome")
```
